```python
with open("bucket_listing.txt", "r") as f:
    lines = f.readlines()

subjects = []

for line in lines:
    if "PRE sub-" in line:
        folder = line.strip().split()[-1].rstrip("/")
        subjects.append(folder)

subjects = sorted(subjects)

print("Number of subject folders:", len(subjects))
print(subjects)
```

```python
repeats = [s for s in subjects if s.endswith("R")]
print("Repeated/replication IDs:", repeats)
```
