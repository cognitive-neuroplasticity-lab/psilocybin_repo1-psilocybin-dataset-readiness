```python
root = Path("/content/ds006072")

with open(root / "dataset_description.json", "r") as f:
    desc = json.load(f)

desc
```

```python
with open(root / "README", "r", encoding="utf-8", errors="ignore") as f:
    readme_text = f.read()

print(readme_text[:3000])
```

```python
with open(root / "CHANGES", "r", encoding="utf-8", errors="ignore") as f:
    changes_text = f.read()

print(changes_text[:2000])
```
